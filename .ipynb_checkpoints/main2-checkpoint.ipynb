```python
import json
import pandas as pd
from sklearn.model_selection import train_test_split

with open("style_descriptions.json", encoding="utf-8") as f:
    desc = json.load(f)

labels = pd.read_csv("house_images/labels.csv")
labels.columns = ["filename", "label"]
labels["text"] = labels["filename"].map(desc)
labels = labels.dropna(subset=["text"]).reset_index(drop=True)

min_count = labels["label"].value_counts().min()

parts = []
for cls in labels["label"].unique():
    part = labels[labels["label"] == cls].sample(n=min_count, random_state=42)
    parts.append(part)

balanced = pd.concat(parts, ignore_index=True)

small, big = train_test_split(
    balanced,
    test_size=0.75,
    stratify=balanced["label"],
    random_state=42,
)
small = small.reset_index(drop=True)
big = big.reset_index(drop=True)

small.to_csv("main_set.csv", index=False)
big.to_csv("best_set.csv", index=False)

print("balanced total:", len(balanced))
print("per class:", min_count)
print("\nmain_set:", len(small))
print(small["label"].value_counts().to_string())
print("\nbest_set:", len(big))
print(big["label"].value_counts().to_string())
```
